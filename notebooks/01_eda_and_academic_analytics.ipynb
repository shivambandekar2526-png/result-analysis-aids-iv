# 📊 Academic Performance & Exploratory Data Analysis (EDA)
**Dataset:** Mumbai University Semester-IV AI & DS Examination Results  
**Batch Size:** 88 Students | **Features:** 113 Columns  


```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv('../data/processed/AI_DS_SEM4_MASTER_RESULTS.csv')
print(f"Dataset Loaded Successfully: {df.shape[0]} students, {df.shape[1]} columns")
df.head(3)
```

### 1. Overall Results Distribution & Summary Statistics

```
print("=== Overall Result Breakdown ===")
print(df['result'].value_counts())
print("\n=== Summary Statistics for Percentage & SGPI ===")
df[['overall_total', 'percentage', 'aCG', 'sgpi']].describe()
```

### 2. SGPA Distribution for Passing Students

```
passed = df[df['result'] == 'SUCCESSFUL']
plt.figure(figsize=(9, 5))
plt.hist(passed['sgpi'], bins=12, color='#2b5c8f', edgecolor='black', alpha=0.85)
plt.axvline(passed['sgpi'].mean(), color='#e74c3c', linestyle='--', linewidth=2, label=f"Mean SGPI: {passed['sgpi'].mean():.2f}")
plt.axvline(passed['sgpi'].median(), color='#f39c12', linestyle=':', linewidth=2, label=f"Median SGPI: {passed['sgpi'].median():.2f}")
plt.title('SGPI Distribution (Successful Students)', fontsize=14, fontweight='bold')
plt.xlabel('SGPI', fontsize=12)
plt.ylabel('Count', fontsize=12)
plt.legend(fontsize=11)
plt.grid(axis='y', alpha=0.5)
plt.show()
```

### 3. Top 10 Rankers in Class

```
toppers = df.sort_values(by=['sgpi', 'overall_total'], ascending=[False, False])
toppers[['seat_no', 'name', 'gender', 'overall_total', 'percentage', 'sgpi', 'result']].head(10).reset_index(drop=True)
```

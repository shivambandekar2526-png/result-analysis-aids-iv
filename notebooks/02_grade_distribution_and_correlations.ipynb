# 📈 Subject Performance, Grade Distributions & Correlations
Analyze subject-level performance, pass/fail trends, and correlations across courses.


```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv('../data/processed/AI_DS_SEM4_MASTER_RESULTS.csv')

subjects = ['OE', 'MINIPROJ', 'CT', 'DBMS', 'OS', 'DBMS_LAB', 'OS_LAB', 'FTS', 'TC_LAB', 'BMD', 'DT']
tot_cols = [f"{s}_total" for s in subjects]
```

### 1. Subject-Wise Average Scores vs Max Marks

```
subj_max = [50, 75, 100, 100, 100, 50, 50, 100, 50, 50, 50]
avg_scores = [df[col].mean() for col in tot_cols]
pct_scores = [a / m * 100 for a, m in zip(avg_scores, subj_max)]

plt.figure(figsize=(12, 5))
plt.bar(subjects, pct_scores, color='#3498db', edgecolor='black', alpha=0.85)
plt.ylabel('Average Percentage (%)', fontsize=12)
plt.title('Subject-wise Average Score (%) Across Class', fontsize=14, fontweight='bold')
plt.ylim(0, 100)
for i, v in enumerate(pct_scores):
    plt.text(i, v + 1.5, f"{v:.1f}%", ha='center', fontweight='bold')
plt.grid(axis='y', alpha=0.5)
plt.show()
```

### 2. Subject-wise Correlation Matrix

```
corr = df[tot_cols].corr()

plt.figure(figsize=(10, 8))
plt.imshow(corr, cmap='coolwarm', interpolation='none', vmin=-0.2, vmax=1.0)
plt.colorbar(label='Pearson Correlation')
plt.xticks(range(len(subjects)), subjects, rotation=45, ha='right')
plt.yticks(range(len(subjects)), subjects)
plt.title('Subject Total Marks Correlation Heatmap', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

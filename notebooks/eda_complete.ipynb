Outliers Detection

```python
Q1 = df["percentage"].quantile(0.25)
Q3 = df["percentage"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[
    (df["percentage"] < lower) |
    (df["percentage"] > upper)
]

print("Upper range:", upper, "\nLower range:", lower)

print(outliers[["name", "percentage", "overall_total"]])
```
